```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python

GDP = pd.read_csv("GDP_Nominal_G20.csv", header=0)


print(GDP[['Reference area','TIME_PERIOD','OBS_VALUE']])
```

```python
GDP = GDP[['Reference area','TIME_PERIOD','OBS_VALUE']]
GDP = GDP.rename(columns={
    'Reference area' : 'Country',
    'TIME_PERIOD' : 'Time',
    'OBS_VALUE': 'Observation value'
})


print(len(GDP['Country'].unique()))
```

```python
print(GDP["Country"].unique())
```

```python

Countries = ['G7','United Kingdom','United States', 'European Union']
GDP = GDP[GDP['Country'].isin(Countries)]
GDP = GDP.sort_values(by='Time')

fig = plt.Figure()
plt.style.use('ggplot')
for country, group in GDP.groupby('Country'):
    plt.plot(group['Time'], group['Observation value'], label=country)
    
plt.title('Real GDP change over time by OECD Country')
plt.xlabel('Time Period')
plt.tick_params('x')
plt.ylabel('GDP Change')
plt.legend(title='Country')
plt.grid(True)
plt.tight_layout()
plt.show()

```
